```python
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd
```

```python
movies_list = {
    "Movie": ['Batman',"Superman","Titanic", "Interstellar", "Pushpa"],
    "Genre": ['Action','Action','Romance','Sci-Fi','Action'],
    "Actor":['Bale','Cavil','DiCaprio','McConaughey','Allu Arjun'],
    "Language": ['English','English','English','English','Telugu']
}

df = pd.DataFrame(movies_list)
df
```

```python
dummies = pd.get_dummies(df.take([1,2,3],axis = 1),dtype = int)
dummies
```

```python
dummies_df = pd.concat([df['Movie'],dummies],axis = 1)
new_dummies_df = dummies_df.set_index(dummies_df.columns[0])
new_dummies_df
```

```python
dict_df = dict(zip(new_dummies_df.index,new_dummies_df.values.tolist()))
dict_df['Batman']
```

```python
dict_df
```

```python
def most_similar_user(target_user, users):

        similar_user = None
        
        max_similarity = float('-inf')

        for user in users:            
            
            if user != target_user:

                similarity = cosine_similarity([users[target_user]],[users[user]])

                if similarity>max_similarity:
                     max_similarity = similarity
                     similar_user = user


        return similar_user


target_user = "Batman"

print(most_similar_user(target_user, dict_df))
```
