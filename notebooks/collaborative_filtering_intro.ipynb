```python
def most_active_user(ratings_df):

    scores = (
        ratings_df
        .sort_values(by='User',ascending=False)
    )
    total_no_of_ratings = ratings_df.count(axis = 1)
    return scores,total_no_of_ratings

print(most_active_user(df))
```
