```python
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd

movies_list = {
    "Movie": ['Batman',"Superman","Titanic", "Interstellar", "Pushpa"],
    "Genre": ['Action','Action','Romance','Sci-Fi','Action'],
    "Actor":['Bale','Cavil','DiCaprio','McConaughey','Allu Arjun'],
    "Language": ['English','English','English','English','Telugu']
}

df = pd.DataFrame(movies_list)
dummies = pd.get_dummies(df.take([1,2,3],axis = 1),dtype = int)
dummies_df = pd.concat([df['Movie'],dummies],axis = 1)
new_dummies_df = dummies_df.set_index(dummies_df.columns[0])
```

```python
similarity_matrix = cosine_similarity(new_dummies_df)
similarity_df = pd.DataFrame(
    similarity_matrix,
    index= new_dummies_df.index,
    columns=new_dummies_df.index
)

similarity_df
```

```python
def recommend(movie_name,top_k):

    try: 
        dct = similarity_df[movie_name].to_dict()
        sorted_data_desc = dict(sorted(dct.items(), key=lambda item: item[1], reverse=True))
        count,index = 0,1

        first_key = next(iter(sorted_data_desc))

        sorted_data_desc.pop(first_key)

        if top_k> len(sorted_data_desc):
            return "Required number of movies not present in the Database."

        for key, value in sorted_data_desc.items():

            if count != top_k:

                print(f'{index}. {key} {value}')
                count += 1
                index += 1
    except (KeyError,ValueError, TypeError):
        print('Unable to Execute the code due to error.')

```

```python
recommend('Superman',3)
```
