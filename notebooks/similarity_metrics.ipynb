```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
import math

def dot_products(a,b):

    dot_product = 0

    for x, y in zip(a,b):
        dot_product += ( x * y)

    return dot_product

def magnitude(values):

    magnitude = 0

    for value in values:

        magnitude += (value) ** 2

    return math.sqrt(magnitude)

def cosine_similarities(a,b):

    dot = dot_products(a,b)

    mag_a = magnitude(a)

    mag_b = magnitude(b)

    cosine_similarity = dot/(mag_a * mag_b)

    return cosine_similarity




# print(cosine_similarities(A,B))

def euclid_distance(a, b):

    total = 0

    for x, y in zip(a,b):

        total += (x-y)**2

    return math.sqrt(total)


A = [1,2]
B = [2,4]

print(cosine_similarities(A,B))
print(cosine_similarity(np.array([A]),np.array([B])))
```

```python

def most_similar_user(target_user, users):

        similar_user = None
        
        max_similarity = 0
        for user in users:            
            
            if user != target_user:

                similarity = cosine_similarities(users[target_user],users[user])

                if similarity>max_similarity:
                     max_similarity = similarity
                     similar_user = user


        return similar_user



users = {
    "Alice": [5,4,5,1],
    "Bob": [4,5,4,1],
    "Charlie": [1,1,5,5],
    "David": [5,5,5,2]
}

target_user = "Alice"

print(most_similar_user(target_user, users))

```
